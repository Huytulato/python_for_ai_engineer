## 1. List là gì?

`list` là một tập hợp **có thứ tự**, **có thể thay đổi** (mutable) các giá trị, có thể chứa nhiều kiểu dữ liệu khác nhau (dù trong thực tế AI/ML thường dùng list chứa cùng 1 kiểu, ví dụ list số `float`).

```python
losses = [0.82, 0.65, 0.41, 0.30, 0.22]
class_names = ["cat", "dog", "bird"]
mixed = ["ResNet50", 50, True]   # được phép nhưng ít dùng trong code ML thực tế

print(losses)
print(class_names)
print(type(losses))
print(len(losses))
```

## 2. Index & Slicing (giống chuỗi)

List dùng index/slicing hoàn toàn tương tự chuỗi — vì chuỗi cũng là một dạng "dãy có thứ tự" (sequence).

```python
losses = [0.82, 0.65, 0.41, 0.30, 0.22]

print(losses[0])     # epoch đầu
print(losses[-1])    # epoch cuối (loss mới nhất)
print(losses[1:3])   # loss của epoch 2 và 3
print(losses[::-1])  # đảo ngược thứ tự
```

## 3. List là **mutable** — sửa đổi trực tiếp

Khác với chuỗi (immutable — không sửa được từng ký tự), list cho phép sửa trực tiếp phần tử, thêm/xoá phần tử.

```python
losses = [0.82, 0.65, 0.41, 0.30, 0.22]

losses[0] = 0.80          # sửa trực tiếp phần tử đầu
print(losses)

losses.append(0.18)       # thêm vào cuối -> giả lập thêm 1 epoch mới
print(losses)

losses.insert(0, 1.0)     # chèn vào vị trí 0
print(losses)

removed = losses.pop()    # xoá và trả về phần tử cuối
print("Đã xoá:", removed, "-> list còn:", losses)

losses.remove(1.0)        # xoá theo giá trị (xoá phần tử đầu tiên có giá trị 1.0)
print(losses)
```

## 4. Các hàm/phương thức hay dùng với list

| Hàm/Phương thức | Tác dụng |
|---|---|
| `len(lst)` | số phần tử |
| `sum(lst)` | tổng các phần tử số |
| `min(lst)` / `max(lst)` | giá trị nhỏ nhất / lớn nhất |
| `lst.sort()` | sắp xếp **tại chỗ** (thay đổi list gốc) |
| `sorted(lst)` | trả về list **mới** đã sắp xếp, giữ list gốc |
| `lst.index(x)` | vị trí đầu tiên chứa giá trị `x` |
| `lst.count(x)` | số lần `x` xuất hiện |
| `x in lst` | kiểm tra `x` có trong list không (`bool`) |

```python
accuracies = [0.72, 0.91, 0.85, 0.91, 0.60]

print("Tổng:", sum(accuracies))
print("Trung bình:", sum(accuracies) / len(accuracies))
print("Cao nhất:", max(accuracies))
print("Thấp nhất:", min(accuracies))

print("Sắp xếp giảm dần (giữ list gốc):", sorted(accuracies, reverse=True))
print("List gốc không đổi:", accuracies)

print("0.91 xuất hiện bao nhiêu lần:", accuracies.count(0.91))
print("0.91 nằm ở vị trí nào (đầu tiên):", accuracies.index(0.91))
print("0.5 có trong list không?", 0.5 in accuracies)
```

## 5. Lặp qua list

Dùng `for` trực tiếp, hoặc `enumerate()` khi cần biết luôn chỉ số (ví dụ: "epoch số mấy ứng với loss này").

```python
losses = [0.82, 0.65, 0.41, 0.30, 0.22]

for loss in losses:
    print(loss)

print("---")

for epoch, loss in enumerate(losses, start=1):   # start=1 để epoch bắt đầu từ 1 thay vì 0
    print(f"Epoch {epoch}: loss = {loss}")
```

## 6. Tuple — "list không đổi được"

`tuple` giống `list` nhưng **immutable** (không thể sửa/thêm/xoá sau khi tạo). Dùng tuple khi dữ liệu là một **nhóm giá trị cố định, không nên bị thay đổi nhầm** — ví dụ: shape của ảnh `(height, width, channels)`, hoặc một cặp tọa độ.

```python
image_shape = (224, 224, 3)   # (height, width, channels) - shape ảnh đầu vào thường gặp

print(image_shape)
print(type(image_shape))
print(image_shape[0])   # vẫn đọc bằng index được

# image_shape[0] = 128   # bỏ comment để tự thấy lỗi: TypeError (tuple không cho sửa)
```

### Unpacking — gán nhiều biến cùng lúc

Rất hay dùng để "bung" một tuple/list ra thành các biến riêng, đọc code rõ nghĩa hơn `shape[0]`, `shape[1]`...

```python
image_shape = (224, 224, 3)
height, width, channels = image_shape

print(f"Ảnh có kích thước {height}x{width}, {channels} kênh màu")

# Unpacking cũng dùng được với list
best_epoch, best_loss = [5, 0.12]
print(best_epoch, best_loss)
```

## 7. List lồng nhau (nested list) — tiền đề của ma trận

Một list có thể chứa các list con bên trong — đây chính là cách biểu diễn **ma trận 2 chiều** trước khi học NumPy (Module 3), nơi mọi thứ sẽ nhanh và tiện hơn rất nhiều.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
]

print(matrix[0])       # dòng đầu tiên: [1, 2, 3]
print(matrix[0][1])    # phần tử dòng 0, cột 1: 2

for row in matrix:
    for value in row:
        print(value, end=" ")
    print()   # xuống dòng sau mỗi hàng
```
