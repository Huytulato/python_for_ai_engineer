## 1. Biến là gì?

Biến giống như một "cái hộp có nhãn" để lưu giá trị, giúp ta dùng lại giá trị đó về sau. Trong Python, bạn **không cần khai báo kiểu dữ liệu** trước — Python tự hiểu kiểu dựa vào giá trị được gán.

```python
name = "Sami"
age = 25
height = 1.75
is_student = True

print(name)
print(age)
print(height)
print(is_student)
```

### Quy tắc đặt tên biến

- Chỉ gồm chữ, số, dấu gạch dưới `_`; **không bắt đầu bằng số**.
- Có phân biệt hoa/thường: `Loss` và `loss` là hai biến khác nhau.
- Không được trùng từ khoá của Python (`class`, `for`, `if`, `return`, ...).
- **Quy ước trong code AI/ML**: dùng `snake_case` (chữ thường, nối bằng `_`), đặt tên có ý nghĩa: `learning_rate`, `batch_size`, `num_epochs`, `y_pred` — không đặt tên mập mờ như `x1`, `tmp2`.

```python
# Hợp lệ
learning_rate = 0.01
batch_size = 32
_hidden_units = 128

print(learning_rate, batch_size, _hidden_units)

# KHÔNG hợp lệ — bỏ comment từng dòng dưới để tự thấy Python báo lỗi gì:
# 2nd_model = "abc"   # không được bắt đầu bằng số
# class = "linear"    # "class" là từ khoá, không dùng làm tên biến
```

## 2. Các kiểu dữ liệu cơ bản

| Kiểu | Ý nghĩa | Ví dụ |
|---|---|---|
| `int` | số nguyên | `10`, `-3` |
| `float` | số thực (có chấm thập phân) | `0.01`, `3.14` |
| `str` | chuỗi ký tự (văn bản) | `"ResNet50"` |
| `bool` | đúng/sai | `True`, `False` |
| `NoneType` | "không có giá trị" | `None` |

Dùng hàm `type(x)` để kiểm tra kiểu dữ liệu của `x`.

```python
num_epochs = 10          # int
loss = 0.0345            # float
model_name = "ResNet50"  # str
is_training = True       # bool
best_model = None        # chưa có model nào -> None

print(num_epochs, type(num_epochs))
print(loss, type(loss))
print(model_name, type(model_name))
print(is_training, type(is_training))
print(best_model, type(best_model))
```

## 3. Ép kiểu dữ liệu (type casting)

Nhiều khi dữ liệu đọc vào ở dạng chuỗi (ví dụ đọc từ file, từ input người dùng) nhưng ta cần tính toán số học → phải ép kiểu bằng `int()`, `float()`, `str()`.

```python
accuracy_str = "0.95"
accuracy = float(accuracy_str)
print(accuracy, type(accuracy))

epoch_str = "10"
epoch = int(epoch_str)
print(epoch, type(epoch))

print(str(42) + "%")   # ghép số với chuỗi phải ép số thành chuỗi trước

# Lỗi thường gặp: int("3.5") sẽ báo ValueError vì "3.5" không phải số nguyên hợp lệ
# Cách đúng: ép qua float trước, rồi mới ép qua int
print(int(float("3.5")))
```

## 4. Toán tử số học

| Toán tử | Ý nghĩa | Ví dụ (`17`, `5`) |
|---|---|---|
| `+` | cộng | `22` |
| `-` | trừ | `12` |
| `*` | nhân | `85` |
| `/` | chia (luôn ra `float`) | `3.4` |
| `//` | chia lấy phần nguyên | `3` |
| `%` | chia lấy phần dư | `2` |
| `**` | lũy thừa | `1419857` |

```python
a = 17
b = 5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

### Vì sao AI Engineer cần nắm chắc `%`, `//`, `**`?

- `//` dùng để tính **số batch** khi chia dữ liệu: `so_batch = so_sample // batch_size`.
- `%` dùng để kiểm tra điều kiện lặp định kỳ: ví dụ "cứ 5 epoch thì lưu checkpoint một lần".
- `**` xuất hiện trực tiếp trong các công thức toán học của ML/DL, ví dụ sai số bình phương `(y_true - y_pred) ** 2`.

```python
# Ví dụ: cứ sau mỗi 5 epoch thì lưu checkpoint
for epoch in range(1, 11):
    if epoch % 5 == 0:
        print(f"Epoch {epoch}: lưu checkpoint")

# Ví dụ: sai số bình phương (squared error) giữa giá trị thật và dự đoán
y_true = 10
y_pred = 8
squared_error = (y_true - y_pred) ** 2
print("Squared error:", squared_error)
```

## 5. Toán tử so sánh & logic

So sánh: `==`, `!=`, `<`, `>`, `<=`, `>=` → luôn trả về `bool` (`True`/`False`).

Logic: `and` (và), `or` (hoặc), `not` (phủ định) — dùng để kết hợp nhiều điều kiện.

```python
x = 10
y = 20

print(x == y)   # bằng nhau?
print(x != y)   # khác nhau?
print(x < y)
print(x >= y)

is_valid = (x < y) and (y > 0)
print(is_valid)

print(not is_valid)
```

## 6. Toán tử gán nhanh

`+=`, `-=`, `*=`, `/=` ... giúp viết ngắn gọn hơn khi cập nhật giá trị của một biến (rất hay dùng khi cộng dồn loss qua các batch).

```python
total_loss = 0
total_loss += 0.5   # tương đương total_loss = total_loss + 0.5
total_loss += 0.3
print(total_loss)

count = 10
count -= 1
print(count)
```

## 7. In dữ liệu với `print()` và f-string

f-string (`f"..."`) cho phép nhúng biến trực tiếp vào chuỗi bằng `{}`, và còn hỗ trợ định dạng số — rất hữu ích khi in loss/accuracy.

```python
model_name = "ResNet50"
accuracy = 0.9456

print(f"Model {model_name} đạt độ chính xác {accuracy:.2%}")   # dạng %, 2 chữ số
print(f"Accuracy làm tròn 2 chữ số thập phân: {accuracy:.4f}")
```
