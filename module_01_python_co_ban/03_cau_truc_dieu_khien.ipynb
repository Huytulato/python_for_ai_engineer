## 1. `if / elif / else`

Cấu trúc rẽ nhánh dựa trên điều kiện đúng/sai (`bool`). Python dùng **thụt lề (indent)** — thường 4 dấu cách — để xác định khối lệnh thuộc về `if` nào, không dùng `{}` như nhiều ngôn ngữ khác.

````python
accuracy = 0.82

if accuracy >= 0.9:
    print("Xuất sắc")
elif accuracy >= 0.7:
    print("Tốt")
elif accuracy >= 0.5:
    print("Trung bình")
else:
    print("Cần cải thiện")
````

### Toán tử điều kiện rút gọn (ternary)

Khi chỉ cần chọn 1 trong 2 giá trị, có thể viết trên 1 dòng: `gia_tri_khi_dung if dieu_kien else gia_tri_khi_sai`.

````python
loss = 0.03
status = "đã hội tụ" if loss < 0.05 else "chưa hội tụ"
print(status)
````

## 2. Vòng lặp `for` với `range()`

`range(start, stop, step)` sinh ra một dãy số từ `start` đến **trước** `stop`. Nếu chỉ truyền 1 số, mặc định `start=0, step=1`.

Đây chính là cấu trúc đứng sau mọi **training loop**: `for epoch in range(num_epochs): ...`

````python
num_epochs = 5

for epoch in range(1, num_epochs + 1):
    print(f"Epoch {epoch}/{num_epochs}")
````

````python
# range với step: ví dụ chỉ log mỗi 2 epoch
for epoch in range(0, 10, 2):
    print("Log tại epoch:", epoch)
````

### Lặp qua chuỗi & `enumerate()`

`for` có thể lặp trực tiếp qua từng ký tự của chuỗi. `enumerate(x)` trả về cả **chỉ số (index)** và **giá trị** cùng lúc — rất hay dùng khi cần biết "đây là batch số mấy".

````python
word = "AI"
for ch in word:
    print(ch)

print("---")

model_name = "GPT"
for idx, ch in enumerate(model_name):
    print(idx, ch)
````

## 3. Vòng lặp `while`

`while` lặp **chừng nào điều kiện còn đúng**. Dùng khi không biết trước số lần lặp — ví dụ: lặp cho đến khi loss giảm dưới một ngưỡng (early stopping).

````python
loss = 1.0
step = 0

while loss > 0.1:
    loss = loss * 0.8   # giả lập loss giảm dần sau mỗi bước
    step += 1
    print(f"Step {step}: loss = {loss:.4f}")

print(f"Dừng sau {step} bước, loss cuối = {loss:.4f}")
````

### `break` và `continue`

- `break`: thoát ngay khỏi vòng lặp.
- `continue`: bỏ qua phần còn lại của lượt lặp hiện tại, nhảy sang lượt kế tiếp.

````python
# break: dừng ngay khi tìm được điều kiện mong muốn
for epoch in range(1, 100):
    loss = 1 / epoch   # giả lập loss giảm dần
    if loss < 0.05:
        print(f"Đạt loss mục tiêu ở epoch {epoch} (loss={loss:.4f}), dừng training.")
        break

# continue: bỏ qua các epoch chẵn, chỉ log epoch lẻ
for epoch in range(1, 8):
    if epoch % 2 == 0:
        continue
    print("Log epoch lẻ:", epoch)
````

## 4. Lồng vòng lặp (nested loop)

Vòng lặp có thể chứa vòng lặp khác bên trong — dùng khi dữ liệu có 2 chiều, ví dụ: lặp qua từng epoch rồi lặp qua từng batch trong epoch đó, hoặc in một bảng số (ma trận) — nền tảng để sau này hiểu NumPy array 2 chiều.

````python
num_epochs = 2
num_batches = 3

for epoch in range(1, num_epochs + 1):
    for batch in range(1, num_batches + 1):
        print(f"Epoch {epoch} - Batch {batch}")
````
