# Bài 5: Dictionary & Set

`dict` là kiểu dữ liệu bạn sẽ dùng **nhiều nhất** khi làm AI Engineer: lưu hyperparameters, lưu config model, lưu kết quả metrics, mapping tên class sang số (`class_to_idx`)... `set` thì hữu ích khi cần loại bỏ trùng lặp hoặc so sánh 2 tập hợp dữ liệu.

Mục tiêu bài học:
- Tạo, truy cập, sửa đổi `dict` (key → value)
- Lặp qua dict bằng `.keys()`, `.values()`, `.items()`
- Dict lồng nhau (nested dict) — mô phỏng file config thực tế
- Tạo và dùng `set`: loại bỏ trùng lặp, các phép toán tập hợp (`union`, `intersection`, `difference`)

> Nhắc lại cách học: đọc lý thuyết → chạy thử ví dụ → tự làm bài tập, đừng xem lời giải trước.

## 1. Dictionary là gì?

`dict` lưu dữ liệu theo cặp **key → value**, không có thứ tự theo chỉ số như list mà truy cập bằng **key**. Rất hợp để lưu config vì mỗi tham số có một "tên" rõ nghĩa.

````python
config = {
    "learning_rate": 0.001,
    "batch_size": 32,
    "num_epochs": 10,
}

print(config)
print(type(config))
print(config["learning_rate"])   # truy cập giá trị qua key
````

## 2. Truy cập an toàn với `.get()`

Nếu truy cập key không tồn tại bằng `dict[key]`, Python sẽ báo lỗi `KeyError`. Dùng `.get(key, default)` để tránh lỗi — trả về `default` nếu key không có.

````python
config = {"learning_rate": 0.001, "batch_size": 32}

print(config.get("batch_size"))            # có key -> trả về giá trị
print(config.get("optimizer"))               # không có key -> trả về None
print(config.get("optimizer", "sgd"))        # không có key -> trả về default "sgd"

# config["optimizer"]   # bỏ comment để tự thấy lỗi KeyError
````

## 3. Thêm / sửa / xoá phần tử

- Gán `dict[key] = value`: nếu key đã có → sửa giá trị; nếu chưa có → thêm mới.
- `.update({...})`: thêm/sửa nhiều cặp cùng lúc.
- `del dict[key]` hoặc `.pop(key)`: xoá một cặp.

````python
config = {"learning_rate": 0.001, "batch_size": 32}

config["optimizer"] = "adam"     # thêm key mới
print(config)

config.update({"batch_size": 64, "num_epochs": 20})   # sửa 1 key có sẵn + thêm 1 key mới
print(config)

del config["num_epochs"]
print(config)

removed_value = config.pop("optimizer")
print("Đã xoá optimizer:", removed_value, "-> config còn:", config)
````

## 4. Kiểm tra key tồn tại & lặp qua dict

````python
config = {"learning_rate": 0.001, "batch_size": 32, "num_epochs": 10}

print("optimizer" in config)        # kiểm tra key có tồn tại
print("learning_rate" in config)

for key in config.keys():           # chỉ lấy key (có thể bỏ .keys(), mặc định for duyệt key)
    print(key)

print("---")
for value in config.values():       # chỉ lấy value
    print(value)

print("---")
for key, value in config.items():   # lấy cả key và value -- dùng nhiều nhất
    print(f"{key}: {value}")
````

## 5. Pattern rất hay dùng: đếm số lần xuất hiện (word count)

Dùng `dict` + `.get(key, 0)` để đếm số lần xuất hiện của mỗi giá trị trong một list — áp dụng để đếm nhãn trong dataset, đếm từ trong văn bản (bước đầu của NLP).

````python
labels = ["cat", "dog", "cat", "bird", "dog", "cat"]

counts = {}
for label in labels:
    counts[label] = counts.get(label, 0) + 1

print(counts)
````

## 6. Dict lồng nhau (nested dict) — mô phỏng file config thực tế

Trong thực tế, config huấn luyện model thường được lưu dưới dạng dict lồng nhau (và thường đọc/ghi từ file JSON — sẽ học ở Module 2).

````python
train_config = {
    "model": {
        "name": "resnet50",
        "num_layers": 50,
    },
    "training": {
        "learning_rate": 0.001,
        "num_epochs": 10,
    },
}

print(train_config["model"]["name"])
print(train_config["training"]["learning_rate"])

train_config["training"]["num_epochs"] = 20   # sửa giá trị lồng sâu bên trong
print(train_config["training"])
````

## 7. Set — tập hợp không trùng lặp

`set` giống `list` nhưng **không giữ thứ tự** và **không cho phần tử trùng lặp**. Rất hữu ích để: lấy danh sách nhãn duy nhất (unique labels) từ dataset, hoặc kiểm tra một giá trị có tồn tại rất nhanh.

````python
labels = ["cat", "dog", "cat", "bird", "dog", "cat"]

unique_labels = set(labels)
print(unique_labels)
print(len(unique_labels), "nhãn khác nhau")

# set cũng tạo trực tiếp bằng {}
colors = {"red", "green", "blue"}
print("red" in colors)   # kiểm tra tồn tại - rất nhanh, nhanh hơn "in list" khi dữ liệu lớn
````

### Các phép toán tập hợp

| Phép toán | Ký hiệu | Ý nghĩa |
|---|---|---|
| Hợp (union) | `a \| b` | phần tử có trong `a` hoặc `b` |
| Giao (intersection) | `a & b` | phần tử có trong **cả hai** |
| Hiệu (difference) | `a - b` | phần tử có trong `a` nhưng **không** có trong `b` |

Dùng rất nhiều khi so sánh 2 tập nhãn, ví dụ: nhãn người dùng nhập vs nhãn model hỗ trợ.

````python
user_labels = {"cat", "dog", "fish"}
model_labels = {"cat", "dog", "bird"}

print("Hợp:", user_labels | model_labels)
print("Giao (model hỗ trợ đúng):", user_labels & model_labels)
print("User có nhưng model KHÔNG hỗ trợ:", user_labels - model_labels)
print("Model hỗ trợ nhưng user chưa dùng:", model_labels - user_labels)
````

## Bài tập thực hành

1. Tạo `config = {"learning_rate": 0.001, "batch_size": 32, "num_epochs": 10}`. In ra từng cặp key-value bằng `.items()`, dạng `"learning_rate: 0.001"`.
2. Từ `config` ở câu 1: thêm key `"optimizer": "adam"`, sau đó dùng `.update()` để sửa `"batch_size"` thành `64`. In `config` cuối cùng.
3. Cho `labels = ["cat", "dog", "cat", "bird", "dog", "cat"]`. Dùng dict + `.get(key, 0)` để đếm số lần xuất hiện mỗi nhãn, in kết quả.
4. Cho `class_names = ["cat", "dog", "bird"]`. Tạo dict `class_to_idx` ánh xạ tên class sang chỉ số (`{"cat": 0, "dog": 1, "bird": 2}`) bằng vòng lặp với `enumerate()`. In kết quả.
5. Cho `user_labels = {"cat", "dog", "fish"}` và `model_labels = {"dog", "fish", "bird"}`. In ra: các nhãn **chung** giữa 2 tập (intersection), và các nhãn user có nhưng model **không hỗ trợ** (difference).
6. **(Nâng cao)** Tạo dict lồng nhau mô phỏng config huấn luyện:
   ```python
   config = {
       "model": {"name": "resnet50", "num_layers": 50},
       "training": {"lr": 0.001, "epochs": 10},
   }
   ```
   In ra câu: `"Model resnet50 (50 lớp) train với lr=0.001 trong 10 epoch"` bằng cách truy cập các key lồng nhau + f-string.

````python
# Bài 1: In config bằng .items()

config = {
    "learning_rate": 0.001,
    "batch_size": 32,
    "num_epochs": 10,
}

for key, value in config.items():
    print(f"{key}: {value}")
````

````python
# Bài 2: Thêm optimizer, update batch_size

config["optimizer"] = "adam"
config.update({"batch_size": 64})
print(config)
````

````python
# Bài 3: Đếm số lần xuất hiện mỗi nhãn

labels = ["cat", "dog", "cat", "bird", "dog", "cat"]
count = {}
for label in labels:
    count[label] = count.get(label, 0) + 1
print(count)
````

````python
# Bài 4: Tạo class_to_idx bằng enumerate
class_names = ["cat", "dog", "bird"]
class_to_idx = {class_name: idx for idx, class_name in enumerate(class_names)}
print(class_to_idx)
````

````python
# Bài 5: Giao và hiệu giữa 2 set nhãn
user_labels = {"cat", "dog", "fish"}
model_labels = {"cat", "dog", "bird"}
print("Hợp:", user_labels | model_labels)
print("Giao (model hỗ trợ đúng):", user_labels & model_labels)
print("User có nhưng model KHÔNG hỗ trợ:", user_labels - model_labels)
print("Model hỗ trợ nhưng user chưa dùng:", model_labels - user_labels)
````

````python
# Bài 6 (nâng cao): Dict lồng nhau
config = {
    "model": {
        "name": "resnet50",
        "num_layers": 50,
    },
    "training": {
        "learning_rate": 0.001,
        "num_epochs": 10,
    },
}
name = config["model"]["name"]
num_layers = config["model"]["num_layers"]
lr = config["training"]["learning_rate"]
epochs = config["training"]["num_epochs"]
print(f"Model {name} ({num_layers} lớp) train với lr={lr} trong {epochs} epoch")
````

## Gợi ý

- Bài 4: `for idx, name in enumerate(class_names): class_to_idx[name] = idx` (nhớ tạo `class_to_idx = {}` trước vòng lặp).
- Bài 6: `config["model"]["name"]`, `config["model"]["num_layers"]`, `config["training"]["lr"]`, `config["training"]["epochs"]`.

## Lời giải mẫu

⚠️ Chỉ chạy ô dưới đây **sau khi** bạn đã tự làm xong 6 bài tập ở trên.

````python
# --- Lời giải mẫu ---

# Bài 1
config = {"learning_rate": 0.001, "batch_size": 32, "num_epochs": 10}
for key, value in config.items():
    print(f"{key}: {value}")

# Bài 2
config["optimizer"] = "adam"
config.update({"batch_size": 64})
print(config)

# Bài 3
labels = ["cat", "dog", "cat", "bird", "dog", "cat"]
counts = {}
for label in labels:
    counts[label] = counts.get(label, 0) + 1
print(counts)

# Bài 4
class_names = ["cat", "dog", "bird"]
class_to_idx = {}
for idx, name in enumerate(class_names):
    class_to_idx[name] = idx
print(class_to_idx)

# Bài 5
user_labels = {"cat", "dog", "fish"}
model_labels = {"dog", "fish", "bird"}
print("Chung:", user_labels & model_labels)
print("User có, model không hỗ trợ:", user_labels - model_labels)

# Bài 6
config2 = {
    "model": {"name": "resnet50", "num_layers": 50},
    "training": {"lr": 0.001, "epochs": 10},
}
name = config2["model"]["name"]
num_layers = config2["model"]["num_layers"]
lr = config2["training"]["lr"]
epochs = config2["training"]["epochs"]
print(f"Model {name} ({num_layers} lớp) train với lr={lr} trong {epochs} epoch")
````
