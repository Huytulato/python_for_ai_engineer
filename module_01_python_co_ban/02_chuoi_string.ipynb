# Bài 2: Chuỗi (String) & Xử lý văn bản

Chuỗi (`str`) là kiểu dữ liệu bạn sẽ chạm vào **mọi ngày** khi làm AI: viết prompt cho LLM, đọc log huấn luyện, xử lý dữ liệu text cho NLP, parse kết quả API...

Mục tiêu bài học:
- Tạo chuỗi, truy cập ký tự bằng index & slicing
- Nắm các phương thức chuỗi dùng nhiều nhất: `strip`, `split`, `join`, `replace`, `lower/upper`, `find`
- Dùng f-string để build prompt template cho LLM
- Hiểu escape character (`\n`, `\t`, ...)

> Nhắc lại cách học: đọc lý thuyết → chạy thử ví dụ → tự làm bài tập, đừng xem lời giải trước.

## 1. Tạo chuỗi

Chuỗi đặt trong `'...'` hoặc `"..."` (không khác nhau), hoặc `"""..."""` cho chuỗi nhiều dòng.

```python
greeting = "Xin chào"
model = 'GPT'
multiline = """Đây là chuỗi
trải trên nhiều dòng."""

print(greeting)
print(model)
print(multiline)
print(len(greeting))   # len() đếm số ký tự trong chuỗi
```

## 2. Escape character (ký tự thoát)

Một số ký tự đặc biệt cần viết bằng dấu `\`:

| Ký tự | Ý nghĩa |
|---|---|
| `\n` | xuống dòng |
| `\t` | tab (thụt lề) |
| `\\` | dấu `\` thật |
| `\"` | dấu `"` thật (trong chuỗi `"..."`) |

Rất hay gặp khi in log huấn luyện nhiều dòng, hoặc build prompt có format cố định.

```python
log = "Epoch 1\tloss=0.52\nEpoch 2\tloss=0.31"
print(log)

path_note = "Đường dẫn: C:\\Users\\model.pt"
print(path_note)
```

## 3. Index & Slicing — truy cập ký tự trong chuỗi

Mỗi ký tự trong chuỗi có một **chỉ số (index)**, bắt đầu từ `0`. Chỉ số âm (`-1`, `-2`, ...) đếm từ cuối chuỗi về.

Cú pháp slicing: `chuoi[start:stop:step]` — lấy từ `start` đến **trước** `stop`, nhảy từng `step` ký tự.

```python
text = "AI Engineer"

print(text[0])     # ký tự đầu: 'A'
print(text[-1])     # ký tự cuối: 'r'
print(text[0:2])    # 'AI'   (lấy từ vị trí 0 đến trước 2)
print(text[3:])      # từ vị trí 3 đến hết: 'Engineer'
print(text[:2])      # từ đầu đến trước vị trí 2: 'AI'
print(text[::-1])    # đảo ngược chuỗi (step = -1)
```

## 4. Các phương thức chuỗi thường dùng

| Phương thức | Tác dụng | Ví dụ |
|---|---|---|
| `.lower()` / `.upper()` | chữ thường / chữ hoa | `"AI".lower()` → `"ai"` |
| `.strip()` | xoá khoảng trắng đầu/cuối | `"  hi  ".strip()` → `"hi"` |
| `.split(sep)` | tách chuỗi thành list | `"a,b,c".split(",")` → `["a","b","c"]` |
| `.join(list)` | nối list thành chuỗi | `"-".join(["a","b"])` → `"a-b"` |
| `.replace(a, b)` | thay `a` bằng `b` | `"AI".replace("A","a")` → `"aI"` |
| `.find(x)` | vị trí xuất hiện của `x` (−1 nếu không có) | `"abc".find("b")` → `1` |
| `.startswith(x)` / `.endswith(x)` | có bắt đầu/kết thúc bằng `x`? | trả về `bool` |

Đây là bộ công cụ **làm sạch dữ liệu text (text cleaning)** — bước đầu tiên của bất kỳ pipeline NLP nào.

```python
raw_text = "   Hello World from AI Engineer   "

cleaned = raw_text.strip().lower()
print(repr(cleaned))   # repr() giúp nhìn rõ khoảng trắng còn sót nếu có

words = cleaned.split()        # split() không truyền tham số -> tách theo khoảng trắng
print(words)
print("Số từ:", len(words))

joined = "_".join(words)
print(joined)

print(cleaned.replace("ai", "AI"))
print(cleaned.find("engineer"))
print(cleaned.startswith("hello"))
```

## 5. Nối chuỗi & lặp chuỗi

Dùng `+` để nối, `*` để lặp. Nối nhiều chuỗi bằng `+` liên tục khá chậm và khó đọc — ưu tiên f-string khi có thể.

```python
a = "AI"
b = "Engineer"
print(a + " " + b)
print("=" * 20)   # hay dùng để in đường kẻ phân cách log
```

## 6. f-string nâng cao — xây prompt template cho LLM

Bạn đã biết f-string cơ bản ở Bài 1. Đây là kỹ năng **cực kỳ quan trọng** khi làm AI Engineer: hầu hết lời gọi LLM API đều cần build một đoạn prompt từ template + biến.

```python
system_role = "trợ lý AI chuyên về machine learning"
question = "Overfitting là gì?"

prompt = f"""Bạn là {system_role}.
Hãy trả lời ngắn gọn câu hỏi sau: {question}"""

print(prompt)

# f-string cũng dùng được để format số, rất hữu ích khi log metric
loss = 0.034521
print(f"[Epoch 3] loss = {loss:.4f}")
```
